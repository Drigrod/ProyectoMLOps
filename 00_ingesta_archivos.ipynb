```python
%sh
# Descomprime el ZIP que subiste al Volume
unzip -o /Volumes/main/loterias_raw/raw_apuestas/apuestas_partitioned.zip \
      -d /Volumes/main/loterias_raw/raw_apuestas/

# Verifica que se crearon las carpetas por fecha
ls -l /Volumes/main/loterias_raw/raw_apuestas/apuestas_partitioned | head
```

```python
#Leer los datos en Spark 
SRC = "/Volumes/main/loterias_raw/raw_apuestas/apuestas_partitioned/"

df = (spark.read
        .option("header", True)
        .csv(SRC))

df.printSchema()
display(df.limit(5))
```

```python
spark.sql("CREATE SCHEMA IF NOT EXISTS main.loterias_bronze")

(df.write
  .format("delta")
  .mode("overwrite")
  .partitionBy("fecha")
  .saveAsTable("main.loterias_bronze.apuestas_bronze"))
```
